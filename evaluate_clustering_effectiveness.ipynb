# Evaluate effectiveness of clustering questions

Evaluate whether the clustering of questions is effective by checking if questions asked in the test are similar to the representative questions selected by clustering.

```python
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
from sklearn.manifold import TSNE
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
# Read representative questions and create their embeddings
with open('output/inv-trig/rep-questions.md', 'r') as rqf:
    rep_questions = rqf.readlines()
len(rep_questions)
```

```python
model = SentenceTransformer("BAAI/bge-m3")
```

```python
embeddings = model.encode(rep_questions)
```

```python
with open('output/inv-trig/test/all-test-questions.md', 'r') as tqf:
    test_questions = tqf.readlines()
len(test_questions)
```

```python
test_embeddings = model.encode(test_questions)
```

```python
closest_questions_info = []

for i, test_embedding in enumerate(test_embeddings):
    similarities = cosine_similarity([test_embedding], embeddings)
    closest_idx = np.argmax(similarities)
    closest_question = rep_questions[closest_idx]

    # Calculate similarity to closest question
    sim_to_closest_question = cosine_similarity(
        [test_embedding], 
        [embeddings[rep_questions.index(closest_question)]]
    )[0][0]

    closest_questions_info.append({
        "Test Question": test_questions[i],
        "Closest Question": closest_question,
        "Similarity to Closest Question": sim_to_closest_question
    })
```

```python
# Display the evaluation results
print("Test Question | Closest Question | Similarity")
for info in closest_questions_info:
    print(f"{info['Test Question'].strip()} | {info['Closest Question']} | {info['Similarity to Closest Question']}")
```
